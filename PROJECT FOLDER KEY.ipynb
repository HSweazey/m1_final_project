### data/  (not on GitHub)
**data/gametime/**  (main data)
- gametime_07_18-09_06.csv: daily resale prices, snapshots Jul 18 - Sep 6
- gametime_09_07-10_21.csv: daily resale prices, snapshots Sep 7 - Oct 21

**data/music.db**  (757 MB)
- SQLite database built by 01_music_db.py from the two Gametime CSVs above

**data/ticketmaster/clean/**
- tm_events_2026-10-19.csv / tm_events_2026-11-02.csv: Ticketmaster pulls,
  all events
- tm_events_priced_2026-10-19.csv / tm_events_priced_2026-11-02.csv: same
  pulls, priced events only
- events.csv: cleaned Ticketmaster shows (5,453)

**data/ticketmaster/interim/**  (saved API answers, reused so lookups never repeat)
- lastfm.jsonl: Last.fm artist lookups
- musicbrainz.jsonl: MusicBrainz artist lookups
- seatgeek_events.jsonl: SeatGeek event lookups
- seatgeek_venues.jsonl: SeatGeek venue lookups
- census_geocoder.jsonl: Census geocoder results for venue addresses

### outputs/  (EDA reports, open in a browser)
- gametime_report.html: auto EDA of gametime_07_18-09_06.csv

- music_db_report.html: auto EDA of data/music.db (both Gametime files combined)
- tixinfo_report.html / tixinfo_merged_report.html: auto EDA of the "tixinfo"
